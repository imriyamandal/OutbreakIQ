```python
# GLOBAL VISUAL SETTINGS

import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(
    style="whitegrid",
    palette="Set2"
)

plt.rcParams["figure.figsize"] = (12, 6)

plt.rcParams["figure.dpi"] = 120

# Font sizes
plt.rcParams["font.size"] = 11

plt.rcParams["axes.titlesize"] = 16

plt.rcParams["axes.labelsize"] = 13

plt.rcParams["xtick.labelsize"] = 11

plt.rcParams["ytick.labelsize"] = 11

# Remove top/right borders
plt.rcParams["axes.spines.top"] = False

plt.rcParams["axes.spines.right"] = False
```

```python
#DISEASE DISTRIBUTION
disease_counts = (
df["disease"]
.value_counts()
)

print(disease_counts)
```

```python
plt.figure(figsize=(12,6))

sns.countplot(
y=df["disease"],
order=disease_counts.index
)

plt.title("Disease Distribution")
plt.xlabel("Count")
plt.ylabel("Disease")
plt.tight_layout()
plt.show()
```

```python
#STATE-WISE CASES
```

```python
df["cases"] = (
    pd.to_numeric(
        df["cases"],
        errors="coerce"
    )
)
state_cases = (
    df.groupby("state_ut")["cases"]
    .sum()
    .sort_values(ascending=False)
)

print(state_cases.head(10))


plt.figure(figsize=(14,8))

sns.barplot(
    x=state_cases.head(10).values,
    y=state_cases.head(10).index
)

plt.title(
    "Top 10 States by Total Cases",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Total Cases",
    fontsize=12
)

plt.ylabel(
    "State",
    fontsize=12
)

plt.tight_layout()

plt.show()
```

```python
#DISTRICT HOTSPOTS
```

```python
district_cases = (
df.groupby("district")["cases"].sum().sort_values(ascending=False))

print(district_cases.head(20))


plt.figure(figsize=(12,8))

sns.barplot(
x=district_cases.head(20).values,
y=district_cases.head(20).index
)

plt.title("Top 20 District Hotspots")
plt.xlabel("Total Cases")
plt.ylabel("District")

plt.show()
```

```python
#YEARLY TREND
```

```python
yearly_cases = (
df.groupby("year")["cases"]
.sum()
)

print(yearly_cases)

plt.figure(figsize=(12,6))

sns.lineplot(
x=yearly_cases.index,
y=yearly_cases.values,
marker="o"
)

plt.title("Yearly Disease Trend")
plt.xlabel("Year")
plt.ylabel("Total Cases")

plt.show()
```

```python
#MONTHLY TREND
```

```python
monthly_cases = (
df.groupby("month")["cases"]
.mean()
)

print(monthly_cases)

plt.figure(figsize=(12,6))

sns.lineplot(
x=monthly_cases.index,
y=monthly_cases.values,
marker="o"
)

plt.title("Monthly Average Cases")
plt.xlabel("Month")
plt.ylabel("Average Cases")

plt.xticks(range(1,13))

plt.show()
```

```python
#SEASONALITY ANALYSIS
```

```python
seasonal_cases = (
    df.groupby(["month", "disease"])["cases"]
    .mean()
    .reset_index()
)

# Month labels
month_labels = [
    "Jan", "Feb", "Mar", "Apr",
    "May", "Jun", "Jul", "Aug",
    "Sep", "Oct", "Nov", "Dec"
]
# PLOT
plt.figure(figsize=(15,8))

sns.lineplot(
    data=seasonal_cases,
    x="month",
    y="cases",
    hue="disease",
    linewidth=2.5,
    marker="o"
)

plt.title(
    "Seasonality of Diseases",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel(
    "Month",
    fontsize=13
)

plt.ylabel(
    "Average Cases",
    fontsize=13
)

plt.xticks(
    ticks=range(1,13),
    labels=month_labels
)

plt.legend(
    title="Disease",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()

plt.show()
```

```python
#CLIMATE CORRELATION
```

```python
climate_cols = [
    "cases",
    "temp",
    "precipitation",
    "lai"
]

corr = (
    df[climate_cols]
    .corr()
)

print(corr)

plt.figure(figsize=(10,7))

sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", linewidths=1,linecolor="white",square=True, cbar_kws={ "shrink": 0.8 }, annot_kws={
        "size": 12,
        "weight": "bold"
    }
)

plt.title(
    "Climate Correlation Heatmap",
    fontsize=18,
    fontweight="bold",
    pad=20
)

plt.xticks(fontsize=11)

plt.yticks(fontsize=11)

plt.tight_layout()

plt.savefig(
    "../reports/climate_correlation_heatmap.png",
    bbox_inches="tight"
)

plt.show()
```

```python
#DEATH RATE ANALYSIS
```

```python
df["death_rate"] = (
df["deaths"] /
(df["cases"] + 1)
)

print(
df["death_rate"]
.describe()
)

plt.figure(figsize=(10,6))

sns.histplot(
df["death_rate"],
bins=40,
kde=True
)

plt.title("Death Rate Distribution")
plt.xlabel("Death Rate")

plt.show()
```

```python
#TOP DEADLY DISEASES
```

```python
deadly = (
df.groupby("disease")["deaths"]
.sum()
.sort_values(ascending=False)
)

print(deadly)

plt.figure(figsize=(12,6))

sns.barplot(
x=deadly.values,
y=deadly.index
)

plt.title("Most Deadly Diseases")
plt.xlabel("Total Deaths")
plt.ylabel("Disease")

plt.show()
```

```python
#STATE-WISE DEATH RATE
```

```python
state_death_rate = (
df.groupby("state_ut")["death_rate"].mean().sort_values(ascending=False))

print(state_death_rate.head(10))

plt.figure(figsize=(12,6))

sns.barplot(
x=state_death_rate.head(10).values,
y=state_death_rate.head(10).index
)

plt.title("Highest Average Death Rate States")
plt.xlabel("Death Rate")
plt.ylabel("State")

plt.show()
```
